#### Data Writing Modes

#### Append
WE allredy have a existiing file in pur folder, append  will just send the copy, just will add another file if we have not selected the name 

Mode	Behavior	Use Case
overwrite	Completely replaces any existing data at the target location with the new DataFrame data.	Ideal for full dataset refreshes where you always want the output to contain only the latest data.
append	Adds the new data from the DataFrame to the existing data at the target location.	Used for incremental loads where new data batches are continuously added to a historical dataset.
errorIfExists (or error)	Throws an exception (error) and aborts the write operation if the target location already contains data.	This is the default mode and is useful for ensuring data integrity and preventing accidental overwrites.
ignore	Silently skips the write operation if the target location already exists. No data is written, and no error is thrown.	Suitable when you want to ensure that a dataset is only created once, preventing subsequent jobs from overwriting or appending to it.

| Mode            | Behavior                                                                                     | Use Case                                                                                                     |
|-----------------|-----------------------------------------------------------------------------------------------|---------------------------------------------------------------------------------------------------------------|
| overwrite       | Completely replaces any existing data at the target location with the new DataFrame data.    | Ideal for full dataset refreshes where you always want the output to contain only the latest data.           |
| append          | Adds the new data from the DataFrame to the existing data at the target location.            | Used for incremental loads where new data batches are continuously added to a historical dataset.            |
| errorIfExists   | Throws an exception (error) and aborts the write operation if the target location already contains data. | Useful for ensuring data integrity and preventing accidental overwrites. (Default mode)                 |
| ignore          | Silently skips the write operation if the target location already exists. No data is written, and no error is thrown. | Suitable when you want to ensure that a dataset is only created once, preventing subsequent jobs from overwriting or appending to it. |
